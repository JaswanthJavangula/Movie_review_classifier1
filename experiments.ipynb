```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
#sentenc

sente = ["the glass of milk", "the glass of juice",
          "the cup of tea", "I am a boy", "I am a girl",
          "the book is interesting", "I like to read books"]
```

```python
sente
```

```python
voc_size = 10000
```

```python
one_hot_repre= [one_hot(word, voc_size) for word in sente]
```

```python
for i in range(len(sente)):
    print(f"Sentence: {sente[i]}")
    print(f"One-hot representation: {one_hot_repre[i]}")
    print()
```

```python
# word embedding representation

from tensorflow.keras.layers import Embedding
# from tensorflow.keras.processing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
set_sent_lent = 8

embedded_docs = pad_sequences(one_hot_repre, padding= "pre",
                              maxlen = set_sent_lent)
embedded_docs
```

```python
# feature representaion
dim = 10

model = Sequential()
model.add(Embedding(voc_size, dim, input_length = set_sent_lent))
model.build(input_shape = (None, set_sent_lent))
model.compile('adam', "mse")
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```

```python
for i in range(len(sente)):
    print(f"Sentence: {sente[i]}")
    print(f"One-hot representation: {one_hot_repre[i]}")
    print(f"Padded representation: {embedded_docs[i]}")
    print()
```
